```python
import pandas as pd

df = pd.DataFrame({
    "Магазин": range(1, 13),
    "Регион": ["Север"]*6 + ["Юг"]*6,
    "Товар": ["А","Б"]*6,
    "Выручка": [120,135,128,142,150,131,138,145,129,136,141,133]})
top3 = df.sort_values("Выручка", ascending=False).head(3)
print(top3[["Магазин","Регион","Выручка"]])
yug = df[(df["Регион"]=="Юг") & (df["Выручка"]>135)]
print("Юг свыше 135:", yug["Магазин"].tolist())
```

```python
product_A_130 = df[(df["Товар"] == "А") & (df["Выручка"] < 130)]
product_A_130
```

3 магазина подошло

```python
product_group = df.groupby("Товар")["Выручка"].agg(["sum","max","min"])
product_group
```

Товар А дает наибольшую выручку
