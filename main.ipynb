Лабораторная работа 1

```python
import pandas as pd

df = pd.DataFrame({
    "Магазин": range(1, 13),
    "Регион": ["Север"]*6 + ["Юг"]*6,
    "Товар": ["А","Б"]*6,
    "Выручка": [120,135,128,142,150,131,138,145,129,136,141,133]})
top3 = df.sort_values("Выручка", ascending=False).head(3)
print(top3[["Магазин","Регион","Выручка"]])
yug = df[(df["Регион"]=="Юг") & (df["Выручка"]>135)]
print("Юг свыше 135:", yug["Магазин"].tolist())
```

```python
product_A_130 = df[(df["Товар"] == "А") & (df["Выручка"] < 130)]
product_A_130
```

3 магазина подошло

```python
product_group = df.groupby("Товар")["Выручка"].agg(["sum","max","min"])
product_group
```

Товар А дает наибольшую выручку

```python
north_group = df[df["Регион"] == "Север"]
north_group.sort_values("Выручка", ascending = False)

```

Лабораторная работа 2

```python
import seaborn as sns

df = pd.Series([20, 35, 28, 40, 55, 30, 38, 45, 25, 36, 42, 33])
sns.boxplot(df)
```

Согласно boxplot выбросов нет — все данные лежат в пределах усов. Границы усов определяются правилом Q1 − 1.5·IQR и Q3 + 1.5·IQR, где IQR = Q3 − Q1. Квартили делят данные на четыре части примерно по 25%: ниже Q1, между Q1 и Q2, между Q2 и Q3, выше Q3. Q1, Q2, Q3 — граничные точки между частями.
